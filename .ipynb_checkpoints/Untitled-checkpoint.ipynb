```python
import pygmt
```

This is a markdown Cell


```python
fig = pygmt.Figure()
fig.basemap(region="g", projection="W15", frame=True)
fig.coast(shorelines=True)
fig.show()
```

```python
fig2 = pygmt.Figure()
fig2.basemap(region="g", projection="W15c", frame=True)
fig2.coast(shorelines=["1/1p,black","2/0.5p,red"])
fig2.show()
fig.coast(shorelines="2/faint,blue")
fig.show()
```

```python
oahu = [-158.3, -157.6, 21.2, 21.8]
oahuFig = pygmt.Figure()
for res in ["crude", "low", "intermediate", "high", "full"]:
    oahuFig.coast(resolution=res, shorelines="1p", region=oahu, projection="M15c")
    oahuFig.coast(land='#667686',water="lightblue")
    oahuFig.coast(shorelines='1/2p,black')
    oahuFig.shift_origin(xshift="15c")
oahuFig.show()
```
